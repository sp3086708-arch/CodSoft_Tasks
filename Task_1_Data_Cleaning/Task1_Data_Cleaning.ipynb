```python
import pandas as pd

df = pd.read_csv('superstore.csv', encoding='latin1')

df.head()
```

```python
print("--- Dataset Information ---")
df.info()

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

print("\n--- Duplicate Rows ---")
print("Total Duplicates:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

if 'Postal_Code' in df.columns:
    df['Postal_Code'] = df['Postal_Code'].fillna(0)

df = df.dropna()

print("Remaining missing values across dataset:", df.isnull().sum().sum())
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d/%m/%Y')
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'], format='%d/%m/%Y')

df['Postal_Code'] = df['Postal_Code'].astype(int)

print("--- Updated Data Types ---")
print(df[['Order_Date', 'Ship_Date', 'Postal_Code', 'Sales']].dtypes)
```

```python
df.to_csv('cleaned_superstore.csv', index=False)

print("Success! Cleaned dataset saved as 'cleaned_superstore.csv'")
```
